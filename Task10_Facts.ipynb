```python
from kanren import Relation, facts, var, run

# Define relations
likes = Relation()
girl = Relation()
color = Relation()
owns = Relation()

# Add facts
facts(likes, ("Ram", "Mango"), ("Bill", "Cindy"))
facts(girl, ("Seema",))
facts(color, ("Rose", "Red"))
facts(owns, ("John", "Gold"))

# Query 1: Ram likes mango
x = var()
print("Ram likes:", run(1, x, likes("Ram", x))[0])

# Query 2: Seema is a girl
print("Seema is a girl:", bool(run(1, x, girl("Seema"))))

# Query 3: Bill likes Cindy
print("Bill likes:", run(1, x, likes("Bill", x))[0])

# Query 4: Rose is red
print("Rose is:", run(1, x, color("Rose", x))[0])

# Query 5: John owns gold
print("John owns:", run(1, x, owns("John", x))[0])
```
